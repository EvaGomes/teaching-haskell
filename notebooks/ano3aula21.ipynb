## Como se escreve uma instrução de filtragem?

Vamos observar melhor as três instruções de filtragem que estivémos a experimentar.

```haskell
maratona |> filter(\elem -> elem |> snd |> (=="Nao terminou."))
```

```haskell
municipios |> filter(\elem -> elem |> distrito |> (=="Lisboa"))
```

```haskell
planetas |> filter(\elem -> elem |> estrela |> (=="Sol"))
```

Com outras cores, vamos salientar as partes de uma **instrução de filtragem**:
<br><br>


Todas as instruções de filtragem:
- começam com **a lista que vamos escolher elementos** ou um identificador da lista,
- seguida de **`|> filter(\elem ->`** (esta parte nunca muda!!),
- seguida de um **teste** que vai ser aplicado a cada elemento da lista, e
- terminam com um **`)`**, que fecha o parênteses aberto a seguir à palavra `filter`.

---
## Como funciona a transformação `filter`?


Vamos aprender como funciona a transformação `filter`, começando por filtrar **listas de números**. Vamos usar a seguinte lista:

```haskell
nums = [21,43,178,35,26]
```

### Exemplo 1

```haskell
nums |> filter(\elem -> elem |> odd)
```

Lista que vai ser filtrada: `nums`
<br>Teste aos elementos: `elem |> odd`

<table>
    <tr style="font-weight:bold"><td>elem</td><td>Teste<br>elem |> odd</td><td>Resultado<br>do teste</td></tr>
    <tr><td>21</td><td>21 |> odd</td><td>True</td></tr>
    <tr><td>43</td><td>43 |> odd</td><td>True</td></tr>
    <tr><td>178</td><td>178 |> odd</td><td>False</td></tr>
    <tr><td>35</td><td>35 |> odd</td><td>True</td></tr>
    <tr><td>26</td><td>26 |> odd</td><td>False</td></tr>
</table>

O resultado vai ser uma **lista que tem apenas os elementos que passaram no teste**:

```haskell
nums |> filter(\elem -> elem |> odd)
```

↳ Esta instrução de filtragem serve para **escolher os elementos ímpares**.

### Exemplo 2

```haskell
nums |> filter(\elem -> elem |> (>1000))
```

Lista que vai ser filtrada: `nums`
<br>Teste aos elementos: `elem |> (>1000)`

<table>
    <tr style="font-weight:bold"><td>elem</td><td>Teste<br>elem |> (>1000)</td><td>Resultado<br>do teste</td></tr>
    <tr><td>21</td><td>21 |> (>1000)</td><td>False</td></tr>
    <tr><td>43</td><td>43 |> (>1000)</td><td>False</td></tr>
    <tr><td>178</td><td>178 |> (>1000)</td><td>False</td></tr>
    <tr><td>35</td><td>35 |> (>1000)</td><td>False</td></tr>
    <tr><td>26</td><td>26 |> (>1000)</td><td>False</td></tr>
</table>

**Nenhum** elemento da lista passou no teste! Por isso, o resultado vai ser uma **lista vazia**:

```haskell
nums |> filter(\elem -> elem |> (>1000))
```

↳ Esta instrução de filtragem serve para **escolher os elementos maiores que 1000**.

### Exemplo 3

```haskell
nums |> filter(\elem -> elem |> (*2) |> (<85) )
```

Lista que vai ser filtrada: `nums`
<br>Teste aos elementos: `elem |> (*2) |> (<85)`

<b>Este teste tem duas transformações!</b><br>Por isso, vamos ter de fazer dois passos até chegar ao resultado de cada teste.

<table>
    <tr style="font-weight:bold"><td>elem</td><td>Teste<br>elem |> (*2) |> (&lt;85)</td><td></td><td>Resultado<br>do teste</td></tr>
    <tr><td>21</td><td>21 |> (*2) |> (&lt;85)</td><td>21➡42➡</td><td>True</td></tr>
    <tr><td>43</td><td>43 |> (*2) |> (&lt;85)</td><td>43➡86➡</td><td>False</td></tr>
    <tr><td>178</td><td>178 |> (*2) |> (&lt;85)</td><td>178➡356➡</td><td>False</td></tr>
    <tr><td>35</td><td>35 |> (*2) |> (&lt;85)</td><td>35➡70➡</td><td>True</td></tr>
    <tr><td>26</td><td>26 |> (*2) |> (&lt;85)</td><td>26➡52➡</td><td>True</td></tr>
</table>

**Nenhum** elemento da lista passou no teste! Por isso, o resultado vai ser uma **lista vazia**:

```haskell
nums |> filter(\elem -> elem |> (*2) |> (<85) )
```

↳ Esta instrução de filtragem serve para **escolher os elementos cujo dobro é menor que 85**.
