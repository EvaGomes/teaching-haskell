## Será que podemos **dar um identificador** a uma lista de strings?

```haskell
festa = ["gomas", "sumo", "rissois", "gomas", "bolo de noz", "batatas fritas", "gomas", "croquetes", "sumo", "folhados", "rissois", "queques", "gomas"]
```

```haskell
festa
```

## E podemos aplicar as **transformações** que já conhecíamos?

```haskell
festa |> length
```

```haskell
festa |> reverse
```

```haskell
festa |> nub
```

```haskell
festa |> sort
```

```haskell
festa |> maximum
```

```haskell
festa |> sum
```

```haskell
festa |> sort |> reverse
```

## E **juntar** com outra lista de strings?

```haskell
festa ++ ["pizza", "arroz doce", "doritos"]
```
