### Inventar um tipo de dados `Aluno`

**1º - Vamos explicar ao Jupyter o que é um `Aluno`**
- Começamos por escrever **`data`**, que é uma palavra especial para o Jupyter, para o avisar que vamos criar um novo tipo. ✨
- Dizemos ao Jupyter este novo tipo se chama **`Aluno`**.
- E ensinamos-lhe que um `Aluno` tem sempre 3 características:
    1. um `nome` que é uma string,
    2. um `núm` que é um número inteiro e
    3. um booleano `gostaDeEspinafres`.

**2º - Vamos criar um `Aluno`**

**3º - Vamos fazer perguntas sobre esse `Aluno`**

E também podemos criar uma **lista com elementos do tipo `Aluno`** e fazer perguntas sobre ela!

```haskell
miniturma = [ Aluno .............. , Aluno .............. , Aluno .............. ]
```

```haskell
miniturma |> length
```

```haskell
miniturma |> last |> nome
```

```haskell
miniturma |> map núm
```
