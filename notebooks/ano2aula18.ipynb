### Junção e Emparelhamento
Completa cada instrução com **`++` ou `#`**.

```haskell
[2,5,3]     [3,2,2,4,1]
```

```haskell
[2,5,3]     [3,2,2,4,1]
```

```haskell
"abacaxi"     "doce"
```

<br>
Qual é o <b>resultado</b> destas instruções?

```haskell
["kiwi","uva","banana"] # ["verde","roxo"]
```

```haskell
["kiwi","uva","banana"] ++ [2,5,3,0]
```

```haskell
[3,2,2,4,1] ++ []
```
