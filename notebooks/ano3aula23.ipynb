Observa a lista `li3`. Sobre cada `Personagem` vemos o nome (`nome`), quantos dedos tem numa mão (`nDedos`) e indicação se é animal ou não (`animal`).

```haskell
data Personagem = Personagem {
    nome :: String, nDedos :: Int, animal :: Bool } deriving (Eq,Ord,Show)
```

```haskell
li3 = [ Personagem "Aladino" 5 False, Personagem "Astrobot" 3 False, Personagem "Barbie" 5 False, Personagem "Bart Simpson" 4 False, Personagem "Batman" 5 False, Personagem "Charmander" 4 False, Personagem "Doctor Doofenshmirtz" 4 False, Personagem "Doraemon" 0 False, Personagem "Godzilla" 4 True, Personagem "Inspetor Max" 4 True, Personagem "Luffy" 5 False, Personagem "Mickey" 4 True, Personagem "Moana" 5 False, Personagem "Nemo" 0 True, Personagem "OliverTsubasa" 5 False, Personagem "Popeye" 5 False, Personagem "Ratchet" 5 True, Personagem "Shrek" 5 False, Personagem "SpongeBob" 4 True]
```
