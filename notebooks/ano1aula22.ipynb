## ➡️ Listas (ou Sequências)

```haskell
[80,35,123,35,18]
```

---
Podemos **colocar espaços** entre as vírgulas e os números?

```haskell
[80, 35, 123, 35, 18]
```

```haskell
[80  ,    35,123, 35,  18]
```

---
O que acontece se nos **esquecermos dos parênteses retos**?

```haskell
80,35,123,35,18
```

---
E se colocarmos **pontos em vez de vírgulas**?

```haskell
[80.35.123.35.18]
```
